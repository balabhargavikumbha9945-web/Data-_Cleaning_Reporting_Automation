```python
import pandas as pd

data = {
    "Order_ID": [1001, 1002, 1003, 1004, 1005, 1006, 1007, 1008, 1008, 1009, 1010, 1011],
    "Date": [
        "01-09-2026", "02-09-2026", "03-09-2026", "04-09-2026",
        "05-09-2026", "06-09-2026", "07-09-2026", "08-09-2026",
        "08-09-2026", "09-09-2026", "10-09-2026", None
    ],
    "Customer": [
        "Ravi", "Priya", "Arun", "Ravi", "Sneha", "Priya",
        "Arun", "Sneha", "Sneha", "Kiran", "Kiran", "Meena"
    ],
    "City": [
        "Hyderabad", "Bangalore", "Chennai", "Hyderabad",
        "Mumbai", "Bangalore", "Chennai", "Mumbai",
        "Mumbai", "Hyderabad", "Hyderabad", "Chennai"
    ],
    "Product": [
        "Laptop", "Mouse", "Keyboard", "Laptop", "Monitor",
        "Mouse", "Keyboard", "Monitor", "Monitor", "laptop",
        "Laptop", "Keyboard"
    ],
    "Quantity": [2, 5, 3, 1, 2, None, 3, 2, 2, 1, 2, 4],
    "Price": [50000, 800, 1500, 50000, 12000, 800, 1500, 12000, 12000, 50000, 50000, 1500],
    "Sales": [100000, 4000, 4500, 50000, 24000, None, 4500, 24000, 24000, 50000, 100000, 6000]
}

df = pd.DataFrame(data)

print("Dataset created successfully!")
df
```

```python
print("Number of rows and columns:")
print(df.shape)

print("\nColumn names:")
print(df.columns.tolist())

print("\nDataset information:")
df.info()
```

```python
print("Missing values in each column:")
print(df.isnull().sum())
```

```python
print("Number of duplicate rows:")
print(df.duplicated().sum())
```

```python
df = df.drop_duplicates()

print("Duplicate records removed.")
print("Number of rows after removing duplicates:", len(df))
```

```python
df["Quantity"] = df["Quantity"].fillna(0)

print("Missing Quantity values handled.")
print(df[["Order_ID", "Quantity"]])
```

```python
df["Sales"] = df["Quantity"] * df["Price"]

print("Sales values calculated successfully.")
print(df[["Order_ID", "Quantity", "Price", "Sales"]])
```

```python
df["Product"] = df["Product"].str.strip().str.title()

print("Product names standardized.")
print(df["Product"].unique())
```

```python
df["Date"] = df["Date"].fillna("Unknown")

print("Missing dates handled.")
print(df[["Order_ID", "Date"]])
```

```python
print("CLEANED DATASET")
display(df)
```

```python
total_orders = df["Order_ID"].nunique()
total_quantity = df["Quantity"].sum()
total_sales = df["Sales"].sum()

print("===== SALES SUMMARY REPORT =====")
print("Total Orders   :", total_orders)
print("Total Quantity :", total_quantity)
print("Total Sales    :", total_sales)
```

```python
product_sales = df.groupby("Product")["Sales"].sum().reset_index()

print("===== PRODUCT-WISE SALES =====")
display(product_sales)
```

```python
city_sales = df.groupby("City")["Sales"].sum().reset_index()

print("===== CITY-WISE SALES =====")
display(city_sales)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))

plt.bar(product_sales["Product"], product_sales["Sales"])

plt.title("Product-wise Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
plt.figure(figsize=(8,5))

plt.bar(city_sales["City"], city_sales["Sales"])

plt.title("City-wise Sales")
plt.xlabel("City")
plt.ylabel("Total Sales")

plt.xticks(rotation=45)
plt.tight_layout()

plt.show()
```

```python
report = pd.DataFrame({
    "Metric": [
        "Total Orders",
        "Total Quantity Sold",
        "Total Sales"
    ],
    "Value": [
        total_orders,
        total_quantity,
        total_sales
    ]
})

print("===== FINAL AUTOMATED REPORT =====")
display(report)
```

```python
df.to_csv("cleaned_sales_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```

```python
report.to_csv("sales_summary_report.csv", index=False)

product_sales.to_csv("product_sales_report.csv", index=False)

city_sales.to_csv("city_sales_report.csv", index=False)

print("All reports saved successfully!")
```
